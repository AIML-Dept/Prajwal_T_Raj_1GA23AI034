

# Easy: Deep-Space Signal Inversion (X, Y, Z Gates)


```python
# Easy: Deep-Space Signal Inversion (X, Y, Z Gates)

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# --- X Gate ---
qc_x = QuantumCircuit(1)
qc_x.x(0)
state_x = Statevector(qc_x)

# --- Y Gate ---
qc_y = QuantumCircuit(1)
qc_y.y(0)
state_y = Statevector(qc_y)

# --- Z Gate ---
qc_z = QuantumCircuit(1)
qc_z.z(0)
state_z = Statevector(qc_z)

print("--- Easy: Signal Inversion Circuits & Results ---")
print("\nX-Gate Circuit:\n", qc_x.draw())
print("X-Gate Statevector:", state_x)

print("\nY-Gate Circuit:\n", qc_y.draw())
print("Y-Gate Statevector:", state_y)

print("\nZ-Gate Circuit:\n", qc_z.draw())
print("Z-Gate Statevector:", state_z)
```

# Medium: Photon Polarization Superposition (H and X Combination)


```python
# Medium: Photon Polarization Superposition (H and X Combination)

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Constructing the beacon state
qc_beacon = QuantumCircuit(1)
qc_beacon.h(0)
qc_beacon.x(0)

final_state = Statevector(qc_beacon)

print("--- Medium: Beacon Superposition ---")
print("Circuit Diagram:\n", qc_beacon.draw())
print("Generated Statevector:", final_state)
print("Is valid quantum state:", final_state.is_valid())
```

# Hard: Atmospheric Turbulence Simulation (Sequence of Five Random Gates)


```python
# Hard: Atmospheric Turbulence Simulation (Sequence of Five Random Gates)

import random
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

random.seed(42)
available_gates = ['x', 'y', 'z', 'h', 's']
chosen_sequence = [random.choice(available_gates) for _ in range(5)]

qc_turb = QuantumCircuit(1)
for gate in chosen_sequence:
    if gate == 'x': qc_turb.x(0)
    elif gate == 'y': qc_turb.y(0)
    elif gate == 'z': qc_turb.z(0)
    elif gate == 'h': qc_turb.h(0)
    elif gate == 's': qc_turb.s(0)

simulated_state = Statevector(qc_turb)

print(f"--- Hard: Atmospheric Turbulence ---")
print(f"Randomly chosen gate sequence: {chosen_sequence}")
print("\nCircuit Diagram:\n", qc_turb.draw())
print("Simulated Final Statevector:", simulated_state)
```

# Real-World: Interplanetary Orbital Decision Protocol (Quantum Coin-Flip)


```python
# Real-World: Interplanetary Orbital Decision Protocol (Quantum Coin-Flip)

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc_coin = QuantumCircuit(1, 1)
qc_coin.h(0)
qc_coin.measure(0, 0)

simulator = AerSimulator()
job = simulator.run(qc_coin, shots=1000)
counts = job.result().get_counts(qc_coin)

total_shots = 1000
prob_zero = counts.get('0', 0) / total_shots
prob_one = counts.get('1', 0) / total_shots

print("--- Real-World: Quantum Coin-Flip ---")
print("Circuit Diagram:\n", qc_coin.draw())
print(f"Measurement Counts (over {total_shots} runs): {counts}")
print(f"Probability of 'Go' (State 0): {prob_zero * 100}%")
print(f"Probability of 'No-Go' (State 1): {prob_one * 100}%")
```

## Challenge: Signal Transformation Identity Proof (HXH = Z)

```python
# Challenge: Signal Transformation Identity Proof (HXH = Z)
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Circuit A: HXH Sequence
qc_hxh = QuantumCircuit(1)
qc_hxh.h(0)
qc_hxh.x(0)
qc_hxh.h(0)
state_hxh = Statevector(qc_hxh)

# Circuit B: Direct Z Gate
qc_z = QuantumCircuit(1)
qc_z.z(0)
state_z = Statevector(qc_z)

are_equivalent = state_hxh.equiv(state_z)

print("--- Challenge: Gate Equivalence Proof ---")
print("HXH Circuit Diagram:\n", qc_hxh.draw())
print("HXH Statevector:", state_hxh)
print("\nDirect Z Circuit Diagram:\n", qc_z.draw())
print("Z-Gate Statevector:", state_z)
print(f"\nAre HXH and Z mathematically equivalent? {are_equivalent}")
```
